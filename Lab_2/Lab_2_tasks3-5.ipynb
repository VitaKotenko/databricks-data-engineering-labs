### 1. Create Medallion Schemas

Create Bronze, Silver, and Gold schemas with managed locations in the personal ADLS container.

```python
import yaml

with open("../Lab_3/config.yml", "r") as file:
    config = yaml.safe_load(file)


catalog = config["catalog"]
source_schema = config["schema"]
storage_account = config["storage_account"]
container = config["container"]
secret_scope = config["scope"]

storage_base = (
    f"abfss://{container}@{storage_account}.dfs.core.windows.net"
)
```

```python
layers = ["bronze", "silver", "gold"]

for layer in layers:
    schema_name = f"{source_schema}_{layer}"
    location = f"{storage_base}/{layer}"

    spark.sql(f"""
        CREATE SCHEMA IF NOT EXISTS {catalog}.{schema_name}
        MANAGED LOCATION '{location}'
    """)

    print(f"Created: {catalog}.{schema_name}")
```
