### 1. Create Medallion Schemas

Create Bronze, Silver, and Gold schemas with managed locations in the personal ADLS container.

```python
CATALOG = "dbr_dev_ua5816bd"
SOURCE_SCHEMA = "mialkovska_viktor594"

STORAGE_BASE = "abfss://kotenkoviktoriia@dlsua5816bd.dfs.core.windows.net"

layers = ["bronze", "silver", "gold"]

for layer in layers:
    schema_name = f"{SOURCE_SCHEMA}_{layer}"
    location = f"{STORAGE_BASE}/{layer}"

    spark.sql(f"""
        CREATE SCHEMA IF NOT EXISTS {CATALOG}.{schema_name}
        MANAGED LOCATION '{location}'
    """)

    print(f"Created: {CATALOG}.{schema_name}")
```
