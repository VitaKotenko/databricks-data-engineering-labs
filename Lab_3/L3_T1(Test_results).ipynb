### Once trigger

The stream processed the available data in a single micro-batch using the `once` trigger.

![зображення_1789126213813.png](./l3_screenshots/зображення_1789126213813.png "зображення_1789126213813.png")

### Conclusion

According to the test results, trigger `availableNow` processed the data using multiple micro-batches, while trigger `once` processed the available data in a single micro-batch.
